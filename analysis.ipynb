```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
data = pd.read_csv('data/makerworld_models.csv')
```

```python
data.head()
```

```python
#return the top 50 models by likes and downloads respectively
top_likes = data.sort_values(by='likes', ascending=False).head(50)
top_downloads = data.sort_values(by='downloads', ascending=False).head(50)
# Plotting the top 50 models by likes
plt.figure(figsize=(12, 6))
plt.bar(top_likes['model_name'], top_likes['likes'], color='blue')
plt.xlabel('Model Name')
plt.ylabel('Likes')
plt.title('Top 50 Models by Likes')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Plotting the top 50 models by downloads
plt.figure(figsize=(12, 6))
plt.bar(top_downloads['model_name'], top_downloads['downloads'], color='green')
plt.xlabel('Model Name')
plt.ylabel('Downloads')
plt.title('Top 50 Models by Downloads')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# show the frequency of each tag in the dataset
tag_counts = data['tags'].str.lower().str.split('|').explode().value_counts()
#show the top 50 tags
top_tags = tag_counts.head(50)
# Plotting the frequency of each tag
plt.figure(figsize=(12, 6))
top_tags.plot(kind='bar', color='orange')
plt.xlabel('Tags')
plt.ylabel('Frequency')
plt.title('Frequency of Each Tag in the Dataset')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show() 
```

```python
#show the next 50 tags
next_tags = tag_counts[50:100]
# Plotting the frequency of each tag
plt.figure(figsize=(12, 6))
next_tags.plot(kind='bar', color='orange')
plt.xlabel('Tags')
plt.ylabel('Frequency')
plt.title('Frequency of Each Tag in the Dataset')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show() 
```

```python
#show the distobution of single color and multi color models using the color_classification column
color_counts = data['color_classification'].value_counts()
# Plotting the distribution of single color and multi color models
plt.figure(figsize=(6, 6))
color_counts.plot(kind='pie', autopct='%1.1f%%', colors=['lightblue', 'lightcoral'])
plt.title('Distribution of Single Color and Multi Color Models')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
#histogram of the number of likes and downloads using bins of 10000
plt.figure(figsize=(12, 6))
plt.hist(data['likes'], bins=range(0, data['likes'].max() + 10000, 10000), color='blue', alpha=0.7, label='Likes')
plt.hist(data['downloads'], bins=range(0, data['downloads'].max() + 10000, 10000), color='green', alpha=0.7, label='Downloads')
plt.xlabel('Number of Likes/Downloads')
plt.ylabel('Frequency')
plt.title('Distribution of Likes and Downloads')
#zoom to only show the range of 0 to 10000 likes and downloads
# plt.xlim(0, 10000)
# plt.ylim(0, 800)
plt.legend()
plt.tight_layout()
plt.show()
```
